```python
dfpj = dfpj[['BODY']].rename(columns={'BODY': 'Text'})
dfpj = dfpj[['Text']]
# dfpj['Text'] = dfpj['Text'].astype(str)  # Convert to string type
dfpj['label'] = 'Grooming'  # Binary label for non-grooming
dfpj = dfpj.drop_duplicates(subset='Text') #removing duplicates
dfpj = dfpj.dropna() #removing null values

import re
# Function to remove symbols not followed by letters
def remove_symbols(text):
    return re.sub(r'[^A-Za-z0-9\s]+(?=\s|$)', '', text)
dfpj['Text'] = dfpj['Text'].apply(remove_symbols)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Initialize TF-IDF Vectorizer
tfidf_vectorizer = TfidfVectorizer(max_features=5000)
X = tfidf_vectorizer.fit_transform(combined_df['Text']).toarray()
Y = combined_df['label']
```

# Data Splitting

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.2, random_state=42
)
```

# Classfication

Naive Bayes

```python
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import classification_report

nb_model = GaussianNB()
nb_model.fit(X_train, Y_train)
y_pred_nb = nb_model.predict(X_test)
print(classification_report(Y_test, y_pred_nb))
```

Logistic Regression

```python
from sklearn.linear_model import LogisticRegression

lr_model = LogisticRegression()
lr_model.fit(X_train, Y_train)
y_pred_lr = lr_model.predict(X_test)
print(classification_report(Y_test, y_pred_lr))
```

SVM

```python
from sklearn.svm import LinearSVC

svm_model = LinearSVC()
svm_model.fit(X_train, Y_train)

y_pred_svm = svm_model.predict(X_test)
print(classification_report(Y_test, y_pred_svm))
```

Predict Function

```python
def predict_text(text_list, vectorizer, model):
    # Preprocess the input text
    processed_texts = [remove_stopwords(remove_symbols(text)) for text in text_list]

    # Transform the text using the trained TF-IDF vectorizer
    text_vectors = vectorizer.transform(processed_texts)

    # Convert the sparse matrix to a dense array if the model requires it
    if isinstance(model, GaussianNB):  # Check if the model is GaussianNB
        text_vectors = text_vectors.toarray()

    # Predict using the trained model
    predictions = model.predict(text_vectors)
    return predictions
```
